```python
from pathlib import Path

# Ruta base en Google Drive
base_path = Path(
    "/content/drive/MyDrive/POSGRADO/2026-1/FUNDAMENTOS DE PROGRAMACIÓN CIENTÍFICA/CLASES/PROYECTO_FINAL/SmartRoot_Proyecto"
)

# Estructura del proyecto
structure = {
    "data": ["raw", "processed"],
    "notebooks": [],
    "src": [
        "__init__.py",
        "sensores.py",
        "indicadores.py",
        "simulacion.py",
        "visualizacion.py"
    ],
    "outputs": ["figuras", "tablas"],
    "docs": ["informe"]
}

# Crear carpeta principal
base_path.mkdir(parents=True, exist_ok=True)

# Crear carpetas y archivos
for folder, contents in structure.items():
    folder_path = base_path / folder
    folder_path.mkdir(parents=True, exist_ok=True)

    for item in contents:
        item_path = folder_path / item

        # Si contiene punto, se asume archivo
        if "." in item:
            item_path.touch(exist_ok=True)
        else:
            item_path.mkdir(parents=True, exist_ok=True)

# Crear README.md
(base_path / "README.md").touch(exist_ok=True)

print(f"\nEstructura creada correctamente en:\n{base_path}")
```
