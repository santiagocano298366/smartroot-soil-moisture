## Paso 3 — Crear la estructura de carpetas

Si las carpetas ya existen, la celda las omite sin errores (`exist_ok=True`).

````python
# Carpetas a crear
carpetas = [
    'data/raw',
    'data/processed',
    'notebooks',
    'src',
    'outputs/figuras',
    'outputs/tablas',
    'docs',
]

creadas  = []
existian = []

for carpeta in carpetas:
    ruta_completa = f"{BASE}/{carpeta}"
    ya_existia = os.path.exists(ruta_completa)
    os.makedirs(ruta_completa, exist_ok=True)
    if ya_existia:
        existian.append(carpeta)
    else:
        creadas.append(carpeta)

if creadas:
    print('✅ Carpetas creadas:')
    for c in creadas:
        print(f'   📁 {c}/')

if existian:
    print('\nℹ️  Carpetas que ya existían (sin cambios):')
    for c in existian:
        print(f'   📁 {c}/')
````

## Paso 4 — Crear archivos base del proyecto

Genera el `README.md` del proyecto y los módulos `.py` vacíos de `src/`  
con su estructura de docstrings lista para completar.

````python
# ── README.md ───────────────────────────────────────────────────────────────
readme_contenido = """# 🌱 SmartRoot

**Sistema portátil de bajo costo para la estimación espacial y en tiempo real
del potencial mátrico del suelo mediante resistividad eléctrica y aprendizaje automático.**

---

## Contexto

Proyecto final de la asignatura Fundamentos de Programación Científica,
Maestría en Automatización y Control Industrial.
Conectado con la tesis de maestría del autor.

## Dataset

Jackisch, C. et al. (2018): Soil moisture and matric potential –
An open field comparison of sensor systems [dataset].
PANGAEA. https://doi.org/10.1594/PANGAEA.892319

Jackisch, C. et al. (2020): Soil moisture and matric potential –
an open field comparison of sensor systems.
Earth System Science Data, 12(1), 683–697.
https://doi.org/10.5194/essd-12-683-2020

## Estructura del proyecto

```
PROYECTO_FINAL/
├── data/
│   ├── raw/          # Archivos originales de PANGAEA (.xlsx)
│   └── processed/    # Datasets limpios (.csv)
├── notebooks/
│   ├── 01_EDA.ipynb
│   ├── 02_programacion_basica.ipynb
│   ├── 03_POO_indicadores.ipynb
│   ├── 04_simulacion_montecarlo.ipynb
│   └── 05_tablero_control.ipynb
├── src/
│   ├── sensores.py
│   ├── indicadores.py
│   ├── simulacion.py
│   └── visualizacion.py
├── outputs/
│   ├── figuras/
│   └── tablas/
├── docs/
└── README.md
```

## Notebooks

| # | Notebook | Contenido |
|---|---|---|
| 01 | EDA | Carga, limpieza y exploración del dataset |
| 02 | Programación básica | Variables, ciclos, condicionales, comprensiones |
| 03 | POO e indicadores | Clases, herencia, KPIs, semáforo |
| 04 | Simulación | Monte Carlo, incertidumbre, ML |
| 05 | Tablero de control | Dashboard, gráficos, visualizaciones |

## Uso

1. Abrir `00_setup_workspace.ipynb` en Google Colab y ejecutar todas las celdas.
2. Subir los archivos `.xlsx` de PANGAEA a `data/raw/`.
3. Ejecutar los notebooks en orden.
"""

ruta_readme = f"{BASE}/README.md"
if not os.path.exists(ruta_readme):
    with open(ruta_readme, 'w', encoding='utf-8') as f:
        f.write(readme_contenido)
    print('✅ README.md creado')
else:
    print('ℹ️  README.md ya existe — no se sobreescribió')
````

````python
# ── Módulos src/ ─────────────────────────────────────────────────────────────
# Cada archivo se crea vacío con su docstring.
# El contenido real se desarrollará en los notebooks correspondientes
# y luego se migrará a estos módulos.

modulos = {
    'sensores.py': '''
"""
sensores.py
===========
Módulo de Programación Orientada a Objetos.
Define la jerarquía de clases para representar sensores de suelo,
mediciones y parcelas agrícolas.

Clases:
    Sensor            -- clase base abstracta
    SensorResistivo   -- sensores basados en resistencia eléctrica (Watermark)
    SensorTDR         -- sensores de reflectometría en dominio del tiempo
    SensorCapacitivo  -- sensores capacitivos (5TM, 10HS)
    SensorTensiometro -- tensiómetros de referencia
    Medicion          -- registro individual de una lectura
    Parcela           -- colección de sensores en una ubicación

Desarrollado en: 03_POO_indicadores.ipynb
"""
''',
    'indicadores.py': '''
"""
indicadores.py
==============
Módulo de indicadores de desempeño (KPIs) para el estado hídrico del suelo.
Implementa el sistema de clasificación tipo semáforo
(Verde / Amarillo / Rojo) para decisiones de riego.

Funciones:
    calcular_kpi_potencial()   -- KPI 1: potencial mátrico promedio (kPa)
    calcular_kpi_variabilidad()-- KPI 2: coeficiente de variación entre sensores
    calcular_kpi_tendencia()   -- KPI 3: tasa de cambio temporal del potencial
    clasificar_semaforo()      -- asigna Verde/Amarillo/Rojo según umbrales
    resumen_indicadores()      -- tabla consolidada de todos los KPIs

Desarrollado en: 03_POO_indicadores.ipynb
"""
''',
    'simulacion.py': '''
"""
simulacion.py
=============
Módulo de simulación de escenarios e incertidumbre.
Implementa la simulación de Monte Carlo para cuantificar
la incertidumbre de medición de los electrodos de bajo costo
y su efecto sobre la estimación del potencial mátrico.

Funciones:
    montecarlo_resistividad()  -- simula ruido de medición (>= 1000 iter)
    simular_escenarios()       -- escenarios: seco / normal / húmedo
    calcular_probabilidad()    -- P(ψ < umbral_riego)
    resumen_montecarlo()       -- estadísticos e intervalos de confianza

Desarrollado en: 04_simulacion_montecarlo.ipynb
"""
''',
    'visualizacion.py': '''
"""
visualizacion.py
================
Módulo de visualización y tablero de control.
Genera todos los elementos visuales del proyecto:
gráficos de barras, líneas, histogramas, mapas de calor
y el tablero de control integrado.

Funciones:
    grafico_series_tiempo()    -- evolución temporal del potencial mátrico
    grafico_barras_kpi()       -- KPIs por zona con colores de semáforo
    histograma_montecarlo()    -- distribución de resultados Monte Carlo
    mapa_calor_parcela()       -- interpolación espacial de la parcela
    diagrama_flujo()           -- flujo del sistema SmartRoot
    tablero_control()          -- dashboard integrado (todos los gráficos)

Desarrollado en: 05_tablero_control.ipynb
"""
''',
}

for nombre, contenido in modulos.items():
    ruta = f"{PATHS['src']}/{nombre}"
    if not os.path.exists(ruta):
        with open(ruta, 'w', encoding='utf-8') as f:
            f.write(contenido.strip())
        print(f'✅ src/{nombre} creado')
    else:
        print(f'ℹ️  src/{nombre} ya existe — no se sobreescribió')
````

````python
# ============================================================
# CORRECCIÓN — define rutas y mueve los archivos subidos
# ============================================================
import shutil, os

# Redefinir la ruta (copia exacta de tu estructura)
BASE     = '/content/drive/MyDrive/POSGRADO/2026-1/FUNDAMENTOS DE PROGRAMACIÓN CIENTÍFICA/CLASES/PROYECTO_FINAL'
DATA_RAW = f'{BASE}/data/raw'

# Archivos que quedaron en /tmp o en el directorio de trabajo de Colab
archivos_esperados = [
    'Psi20.xlsx',
    'Theta20.xlsx',
    'T20.xlsx',
    'vG_JKI_params.xlsx',
    'meteo_jki.xlsx',
]

for archivo in archivos_esperados:
    # Colab los guarda en el directorio de trabajo actual
    origen  = f'/content/{archivo}'
    destino = f'{DATA_RAW}/{archivo}'

    if os.path.exists(origen):
        shutil.move(origen, destino)
        print(f'✅ {archivo}  →  data/raw/')
    elif os.path.exists(destino):
        print(f'ℹ️  {archivo} ya está en data/raw/')
    else:
        print(f'⚠️  {archivo} no encontrado — verifica que se subió correctamente')

# Verificar resultado final
print('\n📂 Contenido de data/raw/:')
for f in sorted(os.listdir(DATA_RAW)):
    tam = os.path.getsize(f'{DATA_RAW}/{f}') / 1024
    print(f'   📊 {f}  ({tam:.0f} KB)')
````
